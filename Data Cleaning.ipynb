```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
data = pd.read_csv('IT_customer_churn.csv')
df = pd.DataFrame(data)
df.insert(0,"CustomerID",range(1,len(df)+1))
df.head()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
total_customer = len(df)
print("Toal Customers : ", total_customer)
```

```python
churned_customer = df[df["Churn"] == "Yes"].shape[0]
print("Churned Customers :", churned_customer)
```

```python
churn_rate = (churned_customer / total_customer)* 100
print("Churn Rate :", round(churn_rate,2),"%")
```

```python
contract_churn = pd.crosstab(df['Contract'], df['Churn'], normalize = "index") * 100 
contract_churn
```

```python
tenure_churn = df.groupby("tenure")["Churn"].apply(lambda x: (x == "Yes").mean()*100)
tenure_churn
```

```python
monthly_charges_churn = df.groupby("Churn")["MonthlyCharges"].mean()
monthly_charges_churn
```

```python
internet_churn = pd.crosstab(df['InternetService'], df['Churn'],normalize="index")*100
internet_churn.round(2)
```

```python
payment_churn = pd.crosstab(df['PaymentMethod'],df['Churn'],normalize='index')*100
payment_churn
```

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"],errors="coerce")
```

```python
churned_revenue = df.loc[df['Churn'] == "Yes","TotalCharges"].sum()
print("Total Charges of churned Customers :", churned_revenue)
```

```python
senior_churn = pd.crosstab(df["SeniorCitizen"],df["Churn"],normalize="index")*100
senior_churn.round(2)
```
